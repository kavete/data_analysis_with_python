# Introduction to ScikitLearn (sklearn)

```python
import pandas as pd
heart_disease = pd.read_csv("heart-disease.csv")
heart_disease.head()
```

```python
# Create feature matrix
X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]
```

```python
# Choose the right model and hyperparameters
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier()

clf.get_params()
```

```python
# Fit the model to the training data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2) 
```

```python
clf.fit(X_train, y_train)
```

```python
# Make a prediction
import numpy as np
y_preds = clf.predict(X_test)
```

```python
y_preds
```

```python
y_test
```

```python
# Evaluate the model
clf.score(X_train, y_train)
```

```python
clf.score(X_test, y_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix,accuracy_score
```

```python
print(classification_report(y_test, y_preds))
```

```python
confusion_matrix(y_test, y_preds)
```

```python
accuracy_score(y_test, y_preds)
```

```python
# Improving the model
# Try different number of n_estimators

np.random.seed(42)

for i in range(10, 100, 10):
    print(f"Trying model with {i} estimators ...")
    clf = RandomForestClassifier(n_estimators=i)
    clf.fit(X_train, y_train)
    print(f"Model accuracy on test set: {(clf.score(X_test, y_test) * 100)} %")
```

```python
import pickle

pickle.dump(clf, open("random_forest_heart_disease_model.pk1", "wb"))
```

```python
loaded_model = pickle.load(open("random_forest_heart_disease_model.pk1", "rb"))
```

```python
loaded_model.score(X_test, y_test)
```
